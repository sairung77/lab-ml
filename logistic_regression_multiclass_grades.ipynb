```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
from matplotlib.lines import Line2D

# ------------------------------
# 1) สร้าง dataset
# ------------------------------
data = {
    "Student": ["S1","S2","S3","S4","S5","S6"],
    "GPA": [3.5, 3.8, 3.2, 3.6, 3.9, 3.4],
    "Lang": [4.0, 2.0, 2.5, 3.8, 2.0, 2.5],
    "CS": [2.5, 4.0, 3.0, 2.8, 3.9, 2.8],
    "Math": [3.0, 3.5, 4.0, 3.0, 3.8, 3.9],
    "Major": ["Language","CS","Math","Language","CS","Math"]
}
df = pd.DataFrame(data)

# ------------------------------
# 2) Pairplot + Boxplot
# ------------------------------
sns.pairplot(df, vars=["Lang","CS","Math"], hue="Major", height=2.5, diag_kind="hist")
plt.suptitle("Scatter Matrix: ความสัมพันธ์ระหว่างเกรดวิชากับสาขา", y=1.02)
plt.show()

for subj in ["Lang", "CS", "Math"]:
    plt.figure(figsize=(8,5))
    sns.boxplot(x="Major", y=subj, data=df, palette="Set2")
    sns.stripplot(x="Major", y=subj, data=df, color="black", alpha=0.6)
    plt.title(f"Distribution ของเกรด {subj} ในแต่ละสาขา")
    plt.show()

# ------------------------------
# 3) Logistic Regression + Decision Boundary
# ------------------------------
# เตรียมข้อมูล
X = df[["Lang","CS","Math"]].values
y = LabelEncoder().fit_transform(df["Major"])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

clf = LogisticRegression(multi_class="ovr", solver="lbfgs", max_iter=1000)
clf.fit(X_train, y_train)

# สร้าง grid สำหรับ Lang vs CS (ถือ Math = ค่าเฉลี่ย)
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))

Z = clf.predict(np.c_[xx.ravel(), yy.ravel(), np.full(xx.ravel().shape, X[:,2].mean())])
Z = Z.reshape(xx.shape)

# plot decision boundary
plt.figure(figsize=(8,6))
plt.contourf(xx, yy, Z, alpha=0.3, cmap="Set2")
scatter = plt.scatter(X[:,0], X[:,1], c=y, cmap="Set2", edgecolor="k", s=100)

# manual legend
classes = ["Language","CS","Math"]
colors = [plt.cm.Set2(i) for i in range(len(classes))]
legend_elements = [Line2D([0], [0], marker='o', color='w', label=cls,
                          markerfacecolor=col, markersize=10)
                   for cls, col in zip(classes, colors)]

plt.legend(handles=legend_elements, title="Major")
plt.xlabel("Lang Grade")
plt.ylabel("CS Grade")
plt.title("Decision Boundary (Lang vs CS)\nถือค่า Math = ค่าเฉลี่ย")
plt.show()
```
