```python
# การ Slice String
# การ slice string คือการดึง บางส่วนของข้อความ จากตำแหน่งที่ต้องการ
# รูปแบบ: string[start:stop:step]
# start: ตำแหน่งเริ่มต้น (เริ่มที่ 0)
# stop: ตำแหน่งสุดท้าย แต่ไม่รวมตำแหน่งนี้
# step: ระยะการก้าวข้าม (ไม่บังคับ)

text = "DataScience"

# ดึงอักขระบางส่วน (start:stop)
print(text[0:4])   # 'Data'  ข้อความตั้งแต่ index 0 ถึง 3 (ไม่รวม 4)
print(text[4:10])  # 'Scienc'

# ย่อรูปแบบสั้นของ slice
print(text[:4])    # 'Data' (จากต้นจนถึงตำแหน่งที่ 4)
print(text[4:])    # 'Science' (จาก index 4 ถึงจบ)
print(text[:])     # 'DataScience' (ทั้งหมด)

# ใช้ index ติดลบ (นับจากท้าย)
print(text[-1])     # 'e' (ตัวสุดท้าย)
print(text[-4:])    # 'ence'
print(text[:-4])    # 'DataSci'

# ใช้ step (ข้ามทีละตัว)
print(text[::2])   # 'DtSine' (ข้ามทีละ 2)
print(text[::-1])  # 'ecneicSataD' (กลับด้านข้อความ)
```

```python
# ตัวอย่างการทำ type casting

# str → int / float
score = "85"
score_int = int(score)     # 85 (int)
score_float = float(score) # 85.0 (float)


# int / float → str
x = 123
y = 45.6
print(str(x))   # "123"
print(str(y))   # "45.6"

# int ↔ float
x = 10
print(float(x))  # 10.0

y = 3.9
print(int(y))    # 3 (ปัดทิ้งส่วนทศนิยม)

# bool() – แปลงเป็นค่าความจริง
print(bool(0))          # False
print(bool(5))          # True
print(bool(""))         # False
print(bool("data"))     # True
print(bool([]))         # False
print(bool([1, 2]))     # True

# list() / set() – แปลงจาก string หรือ iterable
text = "abc"
print(list(text))  # ['a', 'b', 'c']
print(set(text))   # {'a', 'b', 'c'} (ลำดับอาจไม่เรียง)

t = (1, 2, 2, 3)
print(list(t))  # [1, 2, 2, 3]
print(set(t))   # {1, 2, 3}

# dict() – แปลงจาก list ของคู่ key-value
pairs = [("name", "Alice"), ("age", 20)]
person = dict(pairs)  # {'name': 'Alice', 'age': 20}
```

```python
# ตัวอย่างการใช้โมดูล math
import math

# ค่าคงที่ทางคณิตศาสตร์
print(math.pi)     # 3.141592653589793
print(math.e)      # 2.718281828459045

# รากที่สอง / ยกกำลัง
print(math.sqrt(16))    # 4.0
print(math.pow(2, 3))   # 8.0 (ผลลัพธ์เป็น float)

# การปัดเศษทศนิยม
print(math.ceil(3.2))   # 4 (ปัดขึ้น)
print(math.floor(3.8))  # 3 (ปัดลง)
print(math.trunc(3.9))  # 3 (ตัดทศนิยม)

# ค่าสัมบูรณ์และสัญลักษณ์
print(abs(-7.5))
print(math.fabs(-7))     # 7.0 (absolute แบบ float)
print(math.copysign(3, -1))  # -3.0 (ใช้สัญลักษณ์ของตัวที่ 2)

# ล็อกการิทึม
print(math.log(100))      # log ฐาน e (ln)
print(math.log10(100))    # log ฐาน 10
print(math.log2(8))       # log ฐาน 2
```
