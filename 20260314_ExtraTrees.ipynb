## Breast cancer datasets in Scikit-Learn.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
```

```python
X = pd.DataFrame(data["data"], columns=data['feature_names'])
X
```

```python
y = data["target"]
y[:30]
```

## Creating a basic Extra Trees Classifier

```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.model_selection import cross_val_score

etc = ExtraTreesClassifier(random_state=42)
cross_val_score(etc, X, y, cv=5).mean()
```

## Hypertuning

Here are some of the popular tuning parameters provided by Scikit-Learn :-
*   n_estimators : The number of trees in a forest. Default = 100
*   max_depth : The maximum depth of the forest. Default = None i.e the tree will continue to expand untill all nodes are pure
*   min_samples_split : The minimum number of samples required to split a node. Default = 2
*   min_samples_leaf : The minimum number of samples required to be at leaf node

This parameter becomes important specially if max_depth=None. Default = 1.
Here, we will hypertune n_estimators and max_depth by using GridSerchCV

```python
from sklearn.model_selection import GridSearchCV

parameters = {'n_estimators': [50, 100], 'max_depth': [10, 50]}
etc = ExtraTreesClassifier(random_state=42)
clf = GridSearchCV(etc, parameters, cv=3, n_jobs=-1)
clf.fit(X, y)
clf.best_params_
```

```python
clf.best_estimator_
```

```python
clf.best_score_
```

## Feature Importance

```python
best_estimator = clf.best_estimator_
best_estimator.fit(X, y)
feature_importances = pd.Series(
  best_estimator.feature_importances_,
  index=best_estimator.feature_names_in_,
  name='Score')
feature_importances.sort_values(ascending=False)
```

## Comparison

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

import time
```

### Decision Tree

```python
start_time = time.time()

dtc = DecisionTreeClassifier(random_state=42)
cv_score = cross_val_score(dtc, X, y, cv=5).mean()
print("Average Cross Validation Score = ", cv_score)
print("--- %s seconds ---" % (time.time() - start_time))
```

### Random Forest

```python
start_time = time.time()

rfc = RandomForestClassifier(random_state=42)
cv_score = cross_val_score(rfc, X, y, cv=5).mean()
print("Average Cross Validation Score = ", cv_score)
print("--- %s seconds ---" % (time.time() - start_time))
```

### Extra Trees

```python
start_time = time.time()

etc = ExtraTreesClassifier(random_state=42)
cv_score = cross_val_score(etc, X, y, cv=5).mean()
print("Average Cross Validation Score = ", cv_score)
print("--- %s seconds ---" % (time.time() - start_time))
```
