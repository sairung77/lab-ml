```python

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation, PillowWriter

# เวกเตอร์ w (ให้คงที่)
w = np.array([4, 0])
w_unit = w / np.linalg.norm(w)

# เตรียม Figure
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-1, 5)
ax.set_ylim(-1, 5)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.grid(True)
ax.set_title("การเปลี่ยนความยาวเงาตามมุมของ v")
ax.set_xlabel("แกน X")
ax.set_ylabel("แกน Y")

# สร้างเส้นว่างสำหรับ animation
v_line, = ax.plot([], [], 'r-', linewidth=2, label='v')
proj_line, = ax.plot([], [], 'b-', linewidth=2, label='projection of v on w')
w_line, = ax.plot([0, w[0]], [0, w[1]], 'g-', linewidth=2, label='w')
perp_line, = ax.plot([], [], 'k--', linewidth=1)
ax.legend()

# ฟังก์ชันอัพเดทเฟรม
def update(frame):
    angle = np.radians(frame)
    v = np.array([np.cos(angle)*4, np.sin(angle)*4])  # หมุน v ไปรอบๆ

    # คำนวณ projection
    v_proj_len = np.dot(v, w_unit)
    v_proj = v_proj_len * w_unit

    # อัพเดทเส้น
    v_line.set_data([0, v[0]], [0, v[1]])
    proj_line.set_data([0, v_proj[0]], [0, v_proj[1]])
    perp_line.set_data([v[0], v_proj[0]], [v[1], v_proj[1]])
    return v_line, proj_line, perp_line

# สร้าง animation และบันทึกเป็น GIF
ani = FuncAnimation(fig, update, frames=np.linspace(0, 180, 60), interval=100)
ani.save("projection_animation.gif", writer=PillowWriter(fps=10))

print("✅ สร้างไฟล์ชื่อ projection_animation.gif แล้ว เปิดดูได้เลย")
```
