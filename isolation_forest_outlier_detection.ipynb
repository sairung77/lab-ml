```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load an example dataset
df = pd.read_csv('data/Ecommerce Customer Churn.csv')

# Create the pair plot with 'Churn' as hue and set transparency
sns.pairplot(df, hue='Churn', plot_kws={'alpha': 0.5})

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix for numerical columns
correlation_matrix = df.select_dtypes(include=['number']).corr()

# Create the heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create the bar chart
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='PreferedOrderCat', hue='CouponUsed', palette='viridis')
plt.title('Coupon Use by Preferred Order Category')
plt.xlabel('Preferred Order Category')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create the bar chart to show the count of customers in each preferred order category
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='PreferedOrderCat', palette='viridis')
plt.title('Number of Customers by Preferred Order Category')
plt.xlabel('Preferred Order Category')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by 'PreferedOrderCat' and sum 'CouponUsed'
coupon_usage_by_category = df.groupby('PreferedOrderCat')['CouponUsed'].sum().reset_index()

# Create the bar chart
plt.figure(figsize=(12, 7))
sns.barplot(data=coupon_usage_by_category, x='PreferedOrderCat', y='CouponUsed', palette='viridis')
plt.title('Total Coupon Used by Preferred Order Category')
plt.xlabel('Preferred Order Category')
plt.ylabel('Total Coupons Used')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
