# Figure

![](https://matplotlib.org/stable/_images/anatomy.png)

## Figure (รูปทั้งหมด)
The whole figure. The figure is a container of all the child `Axes`, and `Artists` (titles, figure legends, etc), and the canvas. A figure can contain any number of Axes, but will typically have at least one.

## Axes (ระนาบ)
This is what you think of as 'a plot', it is the region of the image with the data space. The Axes contains two (or three in the case of 3D) `Axis` objects **Be aware of the difference between `Axes` and `Axis`**


## Axis (แกน)
These are the number-line-like objects. They take care of setting the graph limits and generating the ticks (the marks on the axis) and ticklabels (strings labeling the ticks).

# The object-oriented interface and the pyplot interface
There are two ways to use Matplotlib:

- Rely on pyplot to automatically create and manage the figures and axes, and use pyplot functions for plotting.
- Explicitly create figures and axes, and call methods on them (the "object-oriented (OO) style").

## Figure

````python
x = np.linspace(0, 5, 100)
y1 = x ** 2
y2 = x ** 3
y3 = x ** 4
````

````python
fig = plt.figure()
# fig.add_axes([left, bottom , width, height])
ax1 = fig.add_axes([0.0, 0.0, 1.0, 1.0])
ax2 = fig.add_axes([0.2, 1, 0.3, 0.3])
ax3 = fig.add_axes([0.5, 0.2, 0.1, 0.3])

ax1.plot(x, y1)
ax1.set_title('Axes 1')

ax2.plot(x, y2)
ax2.set_title('Axes 2')

ax3.plot(x, y3)
ax3.set_title('Axes 3')
````

````python
fig = plt.figure()

# fig.add_axes([left, bottom , width, height])
ax1 = fig.add_axes([0   ,0 ,1 , 1])
ax2 = fig.add_axes([1.1 ,0 ,1 , 1])
ax3 = fig.add_axes([2.2 ,0 ,1 , 1])
ax4 = fig.add_axes([0 ,-1.1 ,1 , 1])
ax5 = fig.add_axes([1.1 ,-1.1 ,1 , 1])

ax1.plot(x, y1)
ax1.set_title('$y_1 = x^2$')
ax1.set_xlabel('x')
ax1.set_ylabel('y1')
ax1.grid()

ax2.plot(x, y2)
ax2.set_title('$y_2 = x^3$')
ax2.set_xlabel('x')
ax2.set_ylabel('y1')
ax2.grid()

# ax3.plot(x, y2)
# ax3.set_title('$y_2 = x^3$')
# ax3.set_xlabel('x')
# ax3.set_ylabel('y1')

ax4.plot(x, y2)
ax4.set_title('$y_2 = x^3$')
ax4.set_xlabel('x')
ax4.set_ylabel('y1')
ax4.grid()

ax5.plot(x, y2)
ax5.set_title('$y_2 = x^3$')
ax5.set_xlabel('x')
ax5.set_ylabel('y1')
ax5.grid()
````

# Figure size, aspect ratio and DPI

[matplotlib.figure](https://matplotlib.org/stable/api/figure_api.html?highlight=figure#module-matplotlib.figure)

**Dots per inches (dpi)** determines how many pixels the figure comprises. The default dpi in matplotlib is 100.

A figure of `figsize=(w,h)` will have
```python
px, py = w*dpi, h*dpi  # pixels
# e.g.
# 6.4 inches * 100 dpi = 640 pixels
```

So in order to obtain a figure with a pixel size of e.g. (1200,600) you may chose several combinations of figure size and dpi, e.g.
```python
figsize=(15,7.5), dpi= 80
figsize=(12,6)  , dpi=100
figsize=( 8,4)  , dpi=150
figsize=( 6,3)  , dpi=200
etc.
```

````python
fig = plt.figure(figsize=(4,3), dpi=100)

# fig.add_axes([left, bottom , width, height])
fig.add_axes([0, 0, 2, 1])
fig.add_axes([2.2, 0, 1, 1])
````

# Color, Line Width, Line Style and Marker
https://matplotlib.org/stable/tutorials/introductory/pyplot.html#controlling-line-properties

````python
x = [1, 2, 3, 4]
y = [1, 2, 3, 4]
````

### Color & Line width

````python
fig, axs = plt.subplots(figsize=(12,3), nrows=1, ncols=4)

axs[0].plot(x, y, color="black")
axs[1].plot(x, y, color="red", linewidth=2)
axs[2].plot(x, y, color="#8B008B", lw=3)
axs[3].plot(x, y, color="green", lw=0.5)
````

### Line Style

````python
fig, axs = plt.subplots(figsize=(12,3), nrows=1, ncols=4)

# possible linestype options are '-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'
axs[0].plot(x, y, color="green", lw=3, linestyle='-')
axs[1].plot(x, y, color="green", lw=3, ls='-.')
axs[2].plot(x, y, color="green", lw=3, ls=':')
axs[3].plot(x, y, color="green", lw=3, ls='--')
````

### Marker

````python
fig, axs = plt.subplots(figsize=(12,4), nrows=1, ncols=4)

# possible marker symbols: marker = '+', 'o', '*', 's', ',', '.', '1', '2', '3', '4', ...
axs[0].plot(x, y, color="blue", lw=1, ls='-', marker='+')
axs[1].plot(x, y, color="blue", lw=0.5, ls='--', marker='o')
axs[2].plot(x, y, color="blue", lw=3, ls='-', marker='s')
axs[3].plot(x, y, color="blue", lw=3, ls='', marker='.', markersize=30)
````

````python
fig, axs = plt.subplots(figsize=(12,3), nrows=1, ncols=4)

# marker size and color
axs[0].plot(x, y, color="purple", lw=1, ls='-', marker='o', markersize=2)
axs[1].plot(x, y, color="purple", lw=1, ls='-', marker='o', markersize=4)
axs[2].plot(x, y, color="green", lw=1, ls='-', marker='o', markersize=8,
            markerfacecolor="red", markeredgecolor="black")
axs[3].plot(x, y, color="purple", lw=1, ls='-', marker='s', markersize=8,
            markerfacecolor="yellow", markeredgewidth=3, markeredgecolor="green")
````

# Axis scales and ticks
Each Axes has two (or three) Axis objects representing the x- and y-axis. These control the scale of the Axis, the tick locators and the tick formatters.

````python
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10, 5))
y = np.random.randn(100)
x = np.arange(len(y))  # make an ordinal for this
y = 10**y
axs[0].plot(x, y)

# "linear", "log", "symlog", "logit", ...
axs[1].set_yscale('log')
axs[1].plot(x, y)
````

# Annotating Figure

https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.annotate.html#matplotlib.axes.Axes.annotate

````python
x = np.arange(10)
y = x ** 2
````

````python
fig, ax = plt.subplots()
ax.scatter(x, y)
ax.annotate('Important!!!', xy=(5, 58))
ax.annotate('hi', xy=(6, 10))
````

````python
x = np.arange(10)
y = x ** 2

fig, ax = plt.subplots()
ax.scatter(x, y)
ax.annotate(
    'mytext',
    xy=(8, 65),                 # หัวลูกศร
    xytext=(2, 50),         # ตำแหน่งตัวอักษร
    arrowprops={'facecolor': 'red', 'shrink': 0.05},
    fontsize=16,
)
````
