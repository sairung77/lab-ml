```python
import matplotlib.pyplot as plt

# Define the data for the pie chart.
# 'months' contains the labels for each slice.
# 'y' contains the size of each slice.
months = ['jan', 'feb', 'march', 'apr', 'may', 'jun']
y = [1.2, 2.8, 2.1, 4, 3, 3.5]  # mockup data

# Define which slices to "explode" or pull out from the center.
# The value is the fraction of the radius with which to offset each slice.
# Here, we are only exploding the second slice (index 1).
expl = (0, 0.15, 0, 0, 0, 0)

# Create the pie chart.
# 'y' is the data for the slices.
# 'explode' uses the expl tuple to offset a slice.
# 'labels' uses the months list for the slice labels.
# 'autopct' formats the percentage text inside each slice.
# 'shadow' adds a shadow effect for a 3D look.
# 'startangle' rotates the starting point of the pie chart.
plt.pie(y, explode=expl, labels=months, autopct='%1.1f%%', shadow=True, startangle=90)



# Display the plot.
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define the data for the plots.
# 'x' is the common x-axis data for all plots.
# 'y1' and 'y2' are the y-axis data for the different plots.
x = [1, 2, 3, 4, 5, 6]
y1 = [1.2, 2.8, 2.1, 4, 3, 3.5]
y2 = [2.2, 1.6, 2.3, 2.0, 2.3, 2.1]

# Set font properties for the plot title and labels.
# This updates the default matplotlib settings.
plt.rc('font', **{'family': 'serif', 'sans-serif': ['Tahoma'], 'size': 15})

# Create a figure with a specified size.
# The figsize argument takes a tuple of (width, height) in inches.
fig, ax = plt.subplots(figsize=(6, 4))

# Add a grid to the plot, only on the y-axis.
# linestyle='--' creates a dashed line.
plt.grid(linestyle='--', axis='y')

# Create the first bar plot for 'y1'.
# 'width' controls the thickness of the bars.
# 'label' is used for the legend.
ax.bar(x, y1, width=0.5, label='Local')

# Create the second bar plot for 'y2', offset from the first.
# The list comprehension `[i + 0.2 for i in x]` shifts the bars to the right.
# 'alpha' makes the bars semi-transparent so you can see overlapping elements.
ax.bar([i + 0.2 for i in x], y2, width=0.5, color='orange', alpha=0.8, label='Export')

# Create a line plot for the 'y1' data.
# 'b' is a shorthand for blue color.
# 'marker' specifies the style of the data points.
# 'markersize' controls the size of the markers.
ax.plot(x, y1, color='b', marker='s', markersize=8) # s for square

# Create a second line plot for the 'y2' data.
# 'r' is a shorthand for red color.
# 'o' specifies the marker style as a circle.
ax.plot(x, y2, color='r', marker='o', markersize=8) # o for circle

# Add labels to the x and y axes.
ax.set_xlabel("Month")
ax.set_ylabel("Sale")

# Add a title to the plot.
ax.set_title("Sales Report")

# Add annotations to the first line plot.
# `zip(x, y1)` pairs up the coordinates.
# `plt.annotate()` adds text to the plot at specified coordinates.
# 'xy' is the position to annotate, 'textcoords' is the coordinate system for the text offset.
# 'xytext' is the text position relative to the annotated point.
# 'ha' (horizontal alignment) centers the text.
for tx, ty in list(zip(x, y1)):
    plt.annotate(
        ty,
        (tx, ty),
        textcoords="offset points",
        xytext=(0, 8),
        ha='center',
        fontsize=12
    )

# Add a legend to the plot to identify the data sets.
ax.legend()

# Set the y-axis limits to ensure data is displayed correctly.
ax.set_ylim(0, 4.5)

# Display the plot.
plt.show()
```

Transforming (mutate) : columns/varaibles



```python
#Assigning multiple columns within the same assign is possible.
df = pd.DataFrame({'temp_c': [17.0, 25.0]},
                  index=['Portland', 'Berkeley'])
df
```

```python
#achieved by directly referencing an existing Series or sequence
df.assign(temp_f = df['temp_c'] * 9 / 5 + 32)
```

```python
#alternative way
df.assign(temp_f=lambda x: x.temp_c * 9 / 5 + 32)
```

Grouping and summarizing (group by)

```python
df = pd.DataFrame({'Animal': ['Falcon', 'Falcon',
                              'Parrot', 'Parrot'],
                   'Max Speed': [380., 370., 24., 26.]})
df
```

```python
df.groupby(['Animal']).max()
```

```python
#Hierarchical Indexes
arrays = [['Falcon', 'Falcon', 'Parrot', 'Parrot'],
          ['Captive', 'Wild', 'Captive', 'Wild']]
index = pd.MultiIndex.from_arrays(arrays, names=('Animal', 'Type'))
df = pd.DataFrame({'Max Speed': [390., 350., 30., 20.]},
                  index=index)
df
```

```python
df.groupby(level=0).mean()
```

```python
df.groupby(level="Type").mean()
```

transform method

```python
index = pd.date_range("10/1/1999", periods=1100)
ts = pd.Series(np.random.normal(0.5, 2, 1100), index)
ts = ts.rolling(window=100, min_periods=100).mean().dropna()
```

```python
ts.head()
```

```python
ts.tail()
```

```python
transformed = ts.groupby(lambda x: x.year).transform(
    lambda x: (x - x.mean()) / x.std()
)
```

```python
grouped = ts.groupby(lambda x: x.year)
```

```python
grouped.mean()
```

```python
grouped.std()
```

```python
# Transformed Data
grouped_trans = transformed.groupby(lambda x: x.year)
```

```python
grouped_trans.mean()
```

```python
grouped_trans.std()
```

```python
compare = pd.DataFrame({"Original": ts, "Transformed": transformed})
compare.plot()
```
